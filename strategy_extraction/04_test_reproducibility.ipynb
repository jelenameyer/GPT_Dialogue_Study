# Testing if GPT-5 with minimal reasoning and temperature = 0 returns sufficiently similar ratings on strategies.

```python
# import packages
import pandas as pd
import matplotlib.cm as cm
import matplotlib.pyplot as plt
```

```python
# reading data
ratings = pd.read_csv("data/ratings.csv", keep_default_na=False, na_values=[""])
compare_data = pd.read_csv("data/ratings_long_to_compare.csv", keep_default_na=False, na_values=[""])
```

```python
# merge data together
merged = ratings.merge(compare_data, on=["conv_id","strategy"], how="outer",
                 suffixes=("_A","_B"), indicator=True)
merged["_merge"].value_counts()
```

```python
both = merged[merged["_merge"] == "both"].copy()
both["agree"] = both["rating_A"] == both["rating_B"]
n_total = len(both)
n_agree = int(both["agree"].sum())
print(f"Identical: {n_agree}/{n_total}  ({n_agree/n_total:.1%})")
print(f"Differ:    {n_total - n_agree}/{n_total}  ({(n_total-n_agree)/n_total:.1%})")
```

```python
pd.crosstab(both["rating_A"], both["rating_B"],
            rownames=["ratings"], colnames=["compare"], margins=True)
```
